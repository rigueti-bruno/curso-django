## Manipulação de Strings com Python

> Strings são cadeias de caracteres de texto. São tratadas como listas/arrays pelo Python.

* Definindo uma string:

```python
texto = 'AULA PYCODEBR'
```

* Fatiando a string:

```python
# Retorna o primeiro caractere:
texto[0]
```

```python
# Retorna o antepenúltimo caractere:
texto[-3]
```

```python
# Retorna o sexto caractere:
texto[5]
```

```python
# Retorna uma parte da string:
texto[5:11]
```

```python
# Retorna a partir do sexto ccaractere até o final:
texto[5:]
```

```python
# Retorna os caracteres antes do quinto caractere:
texto[:4]
```

* Obtendo medidas da string:

```python
# Retorna a quantidade de caracteres da string:
len(texto)
```

```python
# Conta quantas vezes o caractere 'A' aparece na string:
texto.count('A')
```

```python
# Conta quantas vezes a cadeia de caractes 'AULA' aparece na string:
texto.count('AULA')
```

```python
# Conta quantas vezes o caractere 'P' aparece em um determinado segmento da string:
texto.count('P', 5, 11)
```

```python
# Retorna a posição onde um trecho da string começa:
texto.find('AULA')
```

```python
texto.find('PYCODEBR')
```

* Formatando o texto da string:

```python
# Torna todos os caracteres da string maiúsculos:
texto.upper()
```

```python
# Torna todos os caracteres da string minúsculos:
texto.lower()
```

```python
# Torna apenas o primeiro caractere maiúsculo:
texto.capitalize()
```

```python
# Torna os primeiros caracteres de cada palavra maiúsculos:
texto.title()
```

* Manipulando a string:

```python
# Divide a string em uma lista de substrings:
texto.split()
```

```python
# Unindo a lista de substrings em uma única string:
lista_de_palavras = texto.split()
'_'.join(lista_de_palavras)
```

```python
texto2 = '  aula pycodebr  '
```

```python
# Elimina os espaços em branco no início e no fim da string:
texto2.strip()
```

```python
# Elimina os espaços em branco no início da string:
texto2.lstrip()
```

```python
# Elimina os espaços em branco no final da string:
texto2.rstrip()
```
