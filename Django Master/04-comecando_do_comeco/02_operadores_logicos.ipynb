## Operadores Lógicos no Python

```python
a = True
b = False
```

```python
if a and b:
    print('atende à condição')
else:
    print('não atende à condição')
```

```python
if a or b:
    print('atende à condição')
else:
    print('não atende à condição')
```

```python
nome = 'Bruce'
idade = 25
peso = 90
```

```python
if idade == 25:
    print('Idade correta')
else:
    print('Idade incorreta')
```

```python
if idade != 24:
    print('Idade correta')
else:
    print('Idade incorreta')
```

```python
if nome == 'Bruce' and idade == 25:
    print('Dados corretos')
else:
    print('Dados incorretos')
```

```python
if nome == 'Bruce' or idade == 24:
    print('Dados corretos')
else:
    print('Dados incorretos')
```

```python
if (nome == 'Bruce' or idade == 24) and peso == 90:
    print('Dados corretos')
else:
    print('Dados incorretos')
```
