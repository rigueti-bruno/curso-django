## Laços de Repetição no Python

* Laço For: Quando temos uma lista finita de repetições a serem executadas

```python
clientes = ['Ana','Jonas','Felipe','Claudio','Renato']

def envia_email(cliente):
    print(f'Email enviado para o cliente {cliente}')
```

```python
for cliente in clientes:
    envia_email(cliente)
```

* Laço While: Quando não soubermos quantas repetições o laço terá ou o conjunto de elementos for infinito

```python
numero = 0

while numero < 7:
    print(numero)
    numero += 1
```

* Podemos definir o número exato de repetições que o Laço For irá executar com a Função range(n)

```python
# Tabuada de 2:
for i in range(11):
    print(f'2 x {i} = {i * 2}')
```

* Podemos extrair e aproveitar na execução os indices de cada elemento da lista no Laço For com a função enumerate()

```python
for i, cliente in enumerate(clientes):
    print(f'Cliente {i}: {cliente}')
```

* Usamos o *break* para encerrar a execução do Laço quando chegar a um determinado elemento, não executando para todos
> Funciona tanto com o Laço For quanto com o While

```python
for i, cliente in enumerate(clientes):
    if i == 2:
        break
    envia_email(cliente)
```

```python
numero = 0

while numero < 10:
    if numero == 6:
        break
    print(numero)
    numero += 1
```

* Usamos o *continue* para pular a execução do laço quando chegar ao elemento com o indice indicado e executar o seguinte
> Funciona com For e While, mas precisa de atenção ao ser executado com o While para não gerar loop infinito

```python
for i, cliente in enumerate(clientes):
    if i == 2:
        continue
    envia_email(cliente)
```

```python
numero = 0
while numero < 10:
    if numero == 6:
        numero +=1
        continue
    print(numero)
    numero += 1
```
