## Compreensão de Lista (List Comprehension) em Python

```python
numeros = [1,2,3,4,5]
```

* Queremos obter uma nova lista com os valores da lista acima dobrados:

1. Com o Loop For simples:

```python
numeros_dobrados_1 = []

for numero in numeros:
    numeros_dobrados_1.append(numero * 2)

print(numeros_dobrados_1)
```

2. Com Compreensão de Lista usando uma expressão simples:

```python
numeros_dobrados_2 = [n * 2 for n in numeros]
print(numeros_dobrados_2)
```

3. Com Compreensão de Lista usando uma função definida:

```python
def dobro(numero):
    return numero * 2

numeros_dobrados_3 = [dobro(numero) for numero in numeros]
print(numeros_dobrados_3)
```

* Queremos colocar todas as strings em uma Lista em Caixa Alta - todas as letras maiusculas:

```python
nomes = ['Ana','Felipe','João','Arlindo','Carlos']
```

```python
nomes_maiusculos_1 = [nome.upper() for nome in nomes]
print(nomes_maiusculos_1)
```

* Agora queremos colocar em Caixa Alta apenas os nomes da lista que começam com a letra 'A':

> Aplicaremos agora um Condicional à Compreensão de Lista.

```python
nomes_maiusculos_2 = [nome.upper() for nome in nomes if nome[0] == 'A']
print(nomes_maiusculos_2)
```
