## Listas e seus Métodos em Python

* Criando uma Lista Vazia

```python
carros = []
```

* Atribuindo um valor a essa lista (atribui ao final da Lista):

```python
carros.append('Marea')
carros
```

```python
carros.append('Fusca')
carros.append('Chevette')
carros.append('Corcel')
carros.append('Kombi')
carros
```

* Adicionando um elemento em uma posição específica na Lista:

```python
carros.insert(1,'Escort')
carros
```

* Remover o último elemento da Lista:

```python
carros.pop()
carros
```

* Remove o elemento de uma posição especificada:

```python
del carros[1]
carros
```

* Exibir um elemento específico da Lista:

```python
carros[0]
```

* Se não existir elemento da posição desejada, retornará um erro:

```python
carros[4]
```

```python
carros.append('Fusca')
carros
```

* Conta quantas vezes um valor se repete nos elementos da Lista:

```python
carros.count('Fusca')
```

* Reverte a ordem dos elementos da Lista de acordo com suas posições na Lista:

```python
carros.reverse()
carros
```

* Ordena os elementos da Lista em ordem crescente de acordo com seus valores:

```python
carros.sort()
carros
```

```python
numeros = [2,4,3,1,5]
numeros.sort()
numeros
```
