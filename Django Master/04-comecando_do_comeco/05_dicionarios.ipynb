### Dicionários no Python:

* Criando um Dicionário:

```python
meu_dicio = {'nome':'Felipe','idade':25,'profissao':'Dev'}
```

* Formas de Acessar um Elemento do Dicionário:

```python
meu_dicio['nome']
```

```python
meu_dicio.get('profissao')
```

* Removendo um Elemento do Dicionário:

```python
meu_dicio.pop('idade')
```

```python
meu_dicio
```

* Visualizar apenas as Chaves do Dicionário:

```python
meu_dicio.keys()
```

* Visualizar apenas os Valores do Dicionário:

```python
meu_dicio.values()
```

* Excluir/Limpar todos os Elementos do dicionário:

```python
meu_dicio.clear()
```

```python
meu_dicio
```

* Dicionario com Estrutura mais Complexa:

```python
pessoa = {
    'nome': 'Felipe',
    'idade': 25,
    'profissao': 'Dev',
    'interesses': ['Python','Programacao','Notebooks'], # podemos adicionar estruturas de dados complexas como listas ao dicionario
    'pet': {  # podemos adicionar outros dicionarios dentro de um dicionario
        'nome':'Loki',
        'idade':1,
        'peso':'2kg'
    }
}
```

* Acessando os Valores desse dicionário complexo:

```python
# Acessando um valor simples:
pessoa.get('nome')
```

```python
# Acessando uma Lista dentro do dicionario:
pessoa.get('interesses')
```

```python
# Acessando um elemento específico dessa lista:
pessoa.get('interesses')[0]  # Acessando o primeiro interesse
```

```python
# Acessando um Dicionário dentro do dicionario:
pessoa.get('pet')
```

```python
# Acessando um um elemento específico desse dicionário:
pessoa.get('pet').get('nome')  # Acessando o nome do pet
```

* Adicionando Novos pares Chave-Valor ao Dicionário:

```python
# Adicionando um valor simples:
pessoa['ano_nascimento'] = 1997
```

```python
# Adicionando uma Lista:
pessoa['cores_favoritas'] = ['Azul','Preto','Verde']
```

```python
# Adicionando um Dicionário:
pessoa['mae'] = {
    'nome':'Maria',
    'idade':50
}
```

```python
# Acessando dados do dicionario complexo
pessoa
```
