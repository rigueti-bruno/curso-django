## Aplicação da Herança entre Classes no Python

> Herança é uma função do Paradigma da Programação Orientada a Objetos que permite aproveitar em classes novas atributos e métodos de outras funções.

* Classe Mãe:

```python
class Carro:
    # Atributos
    numero_rodas = 4
    quantidade_passageiros = 8
    
    # Métodos
    def acelerar(self):
        print('Acelerando v2...')
        
    def frear(self):
        print('Freando ...')
    
    def buzinar(self):
        print('Buzinando ...')
```

* Classe Filha - Herda da Classe Mãe:

```python
class Uno(Carro):
    modelo = 'Uno'
    marca = 'Fiat'
    ano = '1992'
    
    def farol_milha(self):
        print('Farol de milha ligado!')
```

* Instanciando um Objeto da Classe Filha:

```python
uno = Uno()
```

* Exibindo Elementos Herdados da Classe Mãe:

```python
# Atributo herdado da classe Carro:
print(uno.numero_rodas)
```

```python
# Função herdada da classe Carro:
uno.acelerar()
```

* Exibindo Elementos Novos implementados na Classe Filha:

```python
# Atributo novo da classe Uno:
print(uno.modelo)
```

```python
# Função nova da classe Uno:
uno.farol_milha()
```

* Exibindo valores e resultados de atualizações na Classe Mãe na Classe Filha:

```python
# Atributo herdado atualizado da classe Carro:
print(uno.quantidade_passageiros)
```

```python
# Função herdada atualizada da classe Carro:
uno.acelerar()
```
