## Programação Orientada a Objetos em Python

> **Resumindo**: A Orientação a Objetos é uma forma de representar *coisas* com *código*.

* Exemplo de Definição de Classe de Objetos: Celular

```python
class Celular:
    # Propriedades/Atributos
    marca = 'Nokia'
    modelo = 'Tijolão'
    cor = 'Azul'
    tem_camera = False
    bateria = 'Infinita'
    
    # Funções/Métodos
    def fazer_ligacao(self):
        print('Fazendo ligação...')
    
    def jogar_cobrinha(self):
        print('Jogando cobrinha...')
    
    def despertar(self):
        print('Despertando...')
    
    def calcular(self, v1, v2):
        return v1 + v2
```

* Instanciando um Objeto da Classe Celular:

```python
celular = Celular()
```

* Exibindo Propriedades/Atributos:

```python
# Exibindo a propriedade 'marca':
print(celular.marca)
```

```python
# Exibindo a propriedade 'modelo':
print(celular.modelo)
```

* Execuntando Funções/Métodos:

```python
# Excutando a função 'fazer_ligacao':
celular.fazer_ligacao()
```

```python
# Executando a função 'jogar_cobrinha':
celular.jogar_cobrinha()
```

```python
# Executando a função 'calcular' e atribuindo o resultado à variável 'resultado':
resultado = celular.calcular(5, 10)

# Exibindo o resultado do cálculo:
print(resultado)
```
