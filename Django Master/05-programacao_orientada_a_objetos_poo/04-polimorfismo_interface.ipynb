## Aplicação de Polimorfismo de Interface no Python

> O polimorfismo de interface permite à classe gerar objetos com valores diferentes através de atributos indicados em seu método construtor __init__().

* Definição da Classe Mãe:

```python
class Forma:
    def area(self):
        pass
```

* Criação das Classes Filhas:

```python
class Quadrado(Forma):
    def __init__(self,lado):
        self.lado = lado
    def area(self):
        return self.lado ** 2
```

```python
# Instancia de Classe 1:
quadrado1 = Quadrado(5)
print(quadrado1.area())
```

```python
# Instancia de Classe 2:
quadrado2 = Quadrado(7)
print(quadrado2.area())
```

```python
class Circulo(Forma):
    def __init__(self,raio):
        self.raio = raio
    
    def area(self):
        return 3.14 * self.raio ** 2
```

```python
# Instancia de Classe 3:
circulo1 = Circulo(4)
print(circulo1.area())
```

```python
# Instancia de Classe 4:
circulo2 = Circulo(6)
print(circulo2.area())
```
