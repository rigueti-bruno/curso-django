## Aplicação de Polimorfismo de Classe no Python

> O Polimorfismo de Classe implica em classes diferentes, que herdam da mesma classe mãe, mas que têm formas diferentes de implementação dos atributos e métodos herdados da classe mãe.

* Definindo a Classe Mãe:

```python
class Animal:
    def emitir_som(self):
        print("Qualquer som ...")
```

* Definindo as Classes Filhas:

```python
class Cachorro(Animal):
    def emitir_som(self):
        print("Au Au!")
```

```python
cachorro = Cachorro()
cachorro.emitir_som()
```

```python
class Gato(Animal):
    def emitir_som(self):
        print("Miau!")
```

```python
gato = Gato()
gato.emitir_som()
```
